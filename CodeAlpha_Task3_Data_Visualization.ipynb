# Data Visualization – Sales Data Analysis

## Objective

Transform sales data into clear and meaningful visualizations to identify patterns, trends, comparisons, and insights that support decision-making.

## Key Questions

- Which region generates the highest sales?
- Which category contributes the most to sales?
- Which products have the highest sales?
- How do sales change over time?
- What insights can be identified from the visualizations?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os

print(os.listdir('/content'))
```

```python
import glob

files = glob.glob('/content/*.xlsx')
print(files)
```

```python

df = pd.read_excel('/content/Sale­_project_data.xlsx')

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

print("Columns after removing Unnamed columns:")
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt

region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8, 5))
category_sales.plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
product_sales.plot(kind='bar')

plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

sales_trend = df.groupby('Order Date')['Sales'].sum()

plt.figure(figsize=(10, 5))
sales_trend.plot(kind='line', marker='o')

plt.title('Sales Trend Over Time')
plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(data=df, x='Region', y='Sales')

plt.title('Sales Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Quantity',
    y='Sales',
    hue='Category'
)

plt.title('Sales vs Quantity by Category')
plt.xlabel('Quantity')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Contribution by Category')
plt.tight_layout()
plt.show()
```

## Key Insights and Data Story

### Key Insights

- Regional sales visualization helps identify the differences in sales performance across regions.
- Category-wise analysis shows which categories contribute more to total sales.
- Product-wise visualization helps identify high-performing and low-performing products.
- The sales trend chart shows how sales change over time.
- The box plot helps identify the distribution and unusual sales values across regions.
- The scatter plot shows the relationship between quantity and sales across different categories.

### Data Story

The visualizations provide a clear overview of sales performance across regions, categories, products, and time. These insights can help businesses understand sales patterns, identify important product and category contributions, and support data-driven decision-making.
